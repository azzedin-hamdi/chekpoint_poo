```python

class account:
    def __init__(self, account_number: str , account_holder: str, account_balance : float = 0.0):
        self.account_number : str = account_number
        self.account_balance : float = account_balance
        self.account_holder : str = account_holder
        
    
    def deposit(self, amount : float):
        if amount <= 0 :
            print('Amount must be a positif')
            return 0.0
        
        self.account_balance += amount
        print(f'{amount:.2f} DA desposit successfuly.')
        print(f'Now, Your account balance is : {self.account_balance:.2f} DA')
    
    def withdraw(self, amount : float):
        if amount <= 0 :
            print('Amount must be a positif')
            return 0.0
        
        if self.account_balance > amount:
            self.account_balance -= amount
        else:
            print ("Error, Your account its low !")
        
        print(f'{amount:.2f} DA withdraw successfuly.')
        print(f'Now, Your account balance is : {self.account_balance:.2f} DA')
        
    
    def check_balance(self) :
        return self.account_balance
    
    def display_account(self):
        print('\n========ACCOUNT INFORMATION=========')
        print(f'Account Number is : {self.account_number}')
        print(f'Account holder is : {self.account_holder}')
        print(f'Account balance is : {self.account_balance:.2f} DA')
        print('=====================================')
```

```python
account01 = account('12345678','Azzedine',20000)
```

```python
account01.display_account()
```

```python
account01.deposit(100000)
```

```python
account01.withdraw(5000)
```

```python
print('Your account balance is : ', account01.check_balance())
```
